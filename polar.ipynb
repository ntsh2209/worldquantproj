```python
import polars as pl
```

```python
# creating dat
df = pl.DataFrame({"a": [1, 2, 3], "b": [4, 5, 6]})
df
```

```python
df.shape
```

```python
df["a"]
```

```python
df.select(["a","b"])
```

```python
df = df.with_columns((pl.col("a") +pl.col("b")).alias("c"))
```

```python
df
```

```python
df.group_by("a").agg(pl.col("c").mean().alias("mean_c"))
```

```python
df.sort("c", descending=True)
```

```python
df2 = pl.DataFrame({"a": [1, 2, 3], "d": [7, 8, 9]})
df.join(df2,on="a", how="inner")
```
